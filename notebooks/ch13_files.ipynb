## 13.5 Cleaning up a messy text file: the visitor log

```python
with open("../data/visitor_log.txt", encoding="utf-8") as file:
    raw_lines = file.readlines()

print(len(raw_lines))
for line in raw_lines[:6]:
    print(line, end="")
```

```python
print(raw_lines[0].split("|"))
print(raw_lines[1].split("|"))
```

```python
visits = []
for line in raw_lines:
    line = line.strip()
    if line == "":
        continue
    parts = line.split("|")
    date, time = parts[0].strip().split()
    visitor = parts[1].strip().title()
    reason = parts[2].strip()
    visits.append({"date": date, "time": time,
                   "visitor": visitor, "reason": reason})

print(len(visits))
for visit in visits[:4]:
    print(visit["date"], visit["time"], visit["visitor"], "-", visit["reason"])
```

```python
visit_counts = {}
for visit in visits:
    name = visit["visitor"]
    visit_counts[name] = visit_counts.get(name, 0) + 1

for name in sorted(visit_counts):
    print(f"{name:<15}{visit_counts[name]:>3}")
```

```python
animal_mentions = {}
for visit in visits:
    words = visit["reason"].split()
    if visit["reason"].startswith("returned"):
        animal = words[1]
    elif visit["reason"].startswith("dropped off"):
        continue
    elif visit["reason"].startswith("volunteer"):
        continue
    else:
        animal = words[-1]
    animal_mentions[animal] = animal_mentions.get(animal, 0) + 1

top_animal = ""
top_count = 0
for animal, count in animal_mentions.items():
    if count > top_count:
        top_animal = animal
        top_count = count
print(f"{len(animal_mentions)} different animals were mentioned.")
print(f"Most talked about: {top_animal} ({top_count} visits)")
```

```python
with open("output/visitor_log_clean.txt", "w", encoding="utf-8") as file:
    for visit in visits:
        file.write(f"{visit['date']} {visit['time']} | "
                   f"{visit['visitor']} | {visit['reason']}\n")

clean_text = Path("output/visitor_log_clean.txt").read_text(encoding="utf-8")
for line in clean_text.splitlines()[:4]:
    print(line)
```

## 13.6 CSV files with the csv module

```python
with open("../data/animals.csv", encoding="utf-8") as file:
    for line in file.readlines()[:3]:
        print(line[:60])
```

```python
import csv

with open("../data/animals.csv", encoding="utf-8", newline="") as file:
    reader = csv.DictReader(file)
    animals = list(reader)

print(len(animals))
first = animals[0]
for key, value in first.items():
    print(f"{key:>15}: {value}")
```

```python
print(first["age_years"], type(first["age_years"]))
age = float(first["age_years"])
print(age + 1)
```

```python
status_counts = {}
for animal in animals:
    status = animal["status"]
    status_counts[status] = status_counts.get(status, 0) + 1
print(status_counts)

ages = [float(animal["age_years"]) for animal in animals]
print(f"Average age: {sum(ages) / len(ages):.1f} years")
```

```python
family_dogs = []
for animal in animals:
    if (animal["species"] == "Dog" and animal["status"] == "Available"
            and animal["good_with_kids"] == "yes"):
        family_dogs.append({"id": animal["id"], "name": animal["name"],
                            "breed": animal["breed"],
                            "age_years": animal["age_years"]})

for dog in family_dogs:
    print(dog["id"], dog["name"], dog["breed"], dog["age_years"])
```

```python
with open("output/family_dogs.csv", "w", encoding="utf-8",
          newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["id", "name", "breed",
                                              "age_years"])
    writer.writeheader()
    writer.writerows(family_dogs)

print(Path("output/family_dogs.csv").read_text(encoding="utf-8"))
```

```python
with open("../data/donations.csv", encoding="utf-8", newline="") as file:
    donations = list(csv.DictReader(file))

total = 0
by_method = {}
for donation in donations:
    amount = int(donation["amount"])
    total += amount
    method = donation["method"]
    by_method[method] = by_method.get(method, 0) + amount

print(f"{len(donations)} donations, ${total:,} in total")
for method, amount in by_method.items():
    print(f"  {method:<8}${amount:>6,}")
```

```python
by_month = {}
for donation in donations:
    month = donation["date"][:7]
    by_month[month] = by_month.get(month, 0) + int(donation["amount"])

for month, amount in by_month.items():
    print(f"{month}  ${amount:>5,}")
```

```python
all_skills = set()
for volunteer in data["volunteers"]:
    for skill in volunteer["skills"]:
        all_skills.add(skill)
for skill in sorted(all_skills):
    print(skill)
```
