## 3.9 Putting it together at Sunny Paws

```python
# Daily dry food for Peanut, an adult dog.
# Rule: 15 g of dry food per kg of body weight, rounded.
weight_kg = 28.6
grams_per_kg = 15
daily_food = round(weight_kg * grams_per_kg)
per_meal = daily_food / 2
print(daily_food)
print(per_meal)
```

```python
# Kiwi the beagle is 3. Dogs: 15 human years for the first year,
# 9 more for the second, then 5 per year after that.
dog_age = 3
human_years = 15 + 9 + (dog_age - 2) * 5
print("Kiwi is about " + str(human_years) + " in human years.")
```

```python
# A family adopts two cats and one dog.
dog_fee = 150
cat_fee = 90
total_fee = 2 * cat_fee + dog_fee
print("Total adoption fees: $" + str(total_fee))

# Seniors (age 8 and over) pay half.
senior_dog_fee = dog_fee // 2
print("Senior dog fee: $" + str(senior_dog_fee))
```

```python
# How long has an animal been at the shelter?
days_here = 150
weeks = days_here // 7
extra_days = days_here % 7
print(str(weeks) + " weeks and " + str(extra_days) + " days")
```

```python
# Dry food for 16 adult cats for a week, in kilograms.
cats = 16
grams_per_cat = 60
week_kg = cats * grams_per_cat * 7 / 1000
print(week_kg)
```
