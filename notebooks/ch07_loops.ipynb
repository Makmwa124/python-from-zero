## 7.2 The while loop

```python
minutes = 3
while minutes > 0:
    print(f"Minutes to feeding: {minutes}")
    minutes -= 1
print("Feeding time!")
```

```python
minutes = 5
while minutes > 0:
    print(f"Minutes to feeding: {minutes}")
    minutes -= 1
print("Feeding time!")
```

```python
minutes = 0
while minutes > 0:
    print(f"Minutes to feeding: {minutes}")
    minutes -= 1
print("Feeding time!")
```

```python
bowls = 3
while bowls > 0:
    print(f"{bowls} bowls left to fill")
    bowls -= 1
    print(f"  Filled one. bowls is now {bowls}")
print("All bowls filled.")
```

## 7.3 Infinite loops (and how to stop one)

## 7.6 Loop patterns

### The counter

```python
name = "Bramble"
vowels = 0
for letter in name.lower():
    if letter in "aeiou":
        vowels += 1
print(f"{name} has {vowels} vowels")
```

### The running total

```python
food_per_day = 434
total = 0
for day in range(1, 8):
    total += food_per_day
    print(f"Day {day}: {total:,} g so far")
print(f"Scout eats {total:,} g in a week ({total / 1000} kg)")
```

```python
total = 0
for amount in range(3):
    gift = int(input("Donation in dollars: "))
    total += gift
print(f"Total: ${total}")
```

### Finding the biggest

```python
heaviest = 0
for dog in range(4):
    weight = float(input(f"Weight of dog {dog + 1} in kg: "))
    if weight > heaviest:
        heaviest = weight
print(f"The heaviest dog weighs {heaviest} kg")
```

### The flag

```python
tag = "Kennel 7B"
has_digit = False
for character in tag:
    if character in "0123456789":
        has_digit = True
if has_digit:
    print(f"'{tag}' contains a number")
else:
    print(f"'{tag}' has no numbers")
```

## 7.7 break and continue

```python
name = "Peanut Butter"
for letter in name:
    if letter == " ":
        break
    print(letter, end="")
print()
print("Stopped at the first space.")
```

```python
for day in range(1, 8):
    if day == 1:
        print("Day 1 is Monday: closed")
        continue
    print(f"Day {day}: open 10:00 to 17:00")
```

```python
while True:
    answer = input("Type a pet name (or blank to stop): ")
    if answer == "":
        break
    print(f"Hello, {answer.title()}!")
print("Bye for now.")
```

## 7.8 Keep asking until the answer is valid

```python
species = input("Dog or cat? ").strip().lower()
while species != "dog" and species != "cat":
    print("Sorry, please type dog or cat.")
    species = input("Dog or cat? ").strip().lower()
print(f"Thanks, making a feeding plan for a {species}.")
```

```python
print("42".isdigit())
print("4.5".isdigit())
print("forty".isdigit())
print("".isdigit())
```

```python
while True:
    text = input("How many cans of food? ").strip()
    if text.isdigit():
        cans = int(text)
        break
    print("Please type a whole number, like 12.")
print(f"Recording {cans} cans.")
```

## 7.9 Nested loops: a weekly feeding grid

```python
for day in range(1, 4):
    for meal in range(1, 3):
        print(f"Day {day}, meal {meal}")
```

```python
print("        AM   PM")
for day in range(1, 8):
    print(f"Day {day}", end="  ")
    for meal in range(2):
        print("[ ]", end="  ")
    print()
```

```python
for row in range(1, 5):
    for kennel in range(1, 6):
        number = (row - 1) * 5 + kennel
        print(f"K{number:02}", end=" ")
    print()
```
