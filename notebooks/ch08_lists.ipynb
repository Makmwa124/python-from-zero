## 8.6 Looping over a list

## 8.9 List comprehensions

```python
food = []
for weight in weights:
    food.append(round(weight * 15))
print(food)
```

```python
food = [round(weight * 15) for weight in weights]
print(food)
```

```python
shouting = [name.upper() for name in waiting]
print(shouting)
```

```python
heavy = [weight for weight in weights if weight > 20]
print(heavy)
short_names = [name for name in waiting if len(name) <= 4]
print(short_names)
```

## 8.10 Lists of lists: the kennel grid

```python
kennels = [
    ["Peanut", "Scout", "", "Kiwi", "Daisy"],
    ["Milo", "", "Juniper", "", "Bruno"],
    ["", "Nugget", "", "Nova", ""],
    ["Olive", "", "", "", ""],
]
print(len(kennels))
print(kennels[0])
print(kennels[2][1])
```

```python
kennels[3][1] = "Rusty"
print(kennels[3])
kennels[3][1] = ""
```

```python
for row in kennels:
    for dog in row:
        if dog == "":
            dog = "(empty)"
        print(f"{dog:<9}", end="")
    print()
```

```python
empty = 0
for row in kennels:
    for dog in row:
        if dog == "":
            empty += 1
print(f"{empty} of 20 kennels are free")
```

```python
empty = sum([row.count("") for row in kennels])
print(empty)
```

## 8.12 zip(): walking two lists side by side

```python
waiting = ["Peanut", "Scout", "Kiwi", "Daisy", "Milo"]
weights = [28.6, 28.9, 12.2, 12.2, 10.7]
for name, weight in zip(waiting, weights):
    print(f"{name:<8}{weight:>6.1f} kg")
```

```python
pairs = list(zip(waiting[:3], weights[:3]))
print(pairs)
print(pairs[0])
```

```python
for name, weight in zip(waiting, weights):
    grams = round(weight * 15)
    print(f"{name}: {grams} g a day")
```

```python
large = [name for name, weight in zip(waiting, weights) if weight > 20]
print(large)
```
