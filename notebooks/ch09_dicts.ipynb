## 9.5 Counting with a dictionary

```python
arrivals = ["Dog", "Cat", "Guinea pig", "Rabbit", "Cat", "Guinea pig",
            "Dog", "Dog", "Dog", "Dog", "Rabbit", "Rabbit"]
counts = {}
for species in arrivals:
    if species in counts:
        counts[species] += 1
    else:
        counts[species] = 1
print(counts)
```

```python
counts = {}
for species in arrivals:
    counts[species] = counts.get(species, 0) + 1
print(counts)
```

```python
methods = "online online card cash cash card cash cash cash online".split()
tally = {}
for method in methods:
    tally[method] = tally.get(method, 0) + 1
for method, count in tally.items():
    print(f"{method:<7}{'#' * count} {count}")
```

## 9.7 A list of dictionaries: the shelter's animals

```python
animals = [
    {"id": "SP-001", "name": "Peanut", "species": "Dog", "age": 4.5,
     "weight": 28.6, "status": "Available", "good_with_kids": False},
    {"id": "SP-002", "name": "Truffle", "species": "Cat", "age": 1,
     "weight": 3.9, "status": "Medical hold", "good_with_kids": True},
    {"id": "SP-013", "name": "Toast", "species": "Cat", "age": 3.5,
     "weight": 4.0, "status": "Available", "good_with_kids": False},
    {"id": "SP-016", "name": "Ollie", "species": "Cat", "age": 14,
     "weight": 3.9, "status": "Available", "good_with_kids": False},
    {"id": "SP-019", "name": "Bramble", "species": "Rabbit", "age": 2,
     "weight": 1.4, "status": "Available", "good_with_kids": True},
    {"id": "SP-031", "name": "Rusty", "species": "Guinea pig", "age": 1,
     "weight": 1.2, "status": "Available", "good_with_kids": True},
    {"id": "SP-037", "name": "Juniper", "species": "Dog", "age": 10,
     "weight": 22.3, "status": "Available", "good_with_kids": True},
    {"id": "SP-044", "name": "Olive", "species": "Dog", "age": 0.9,
     "weight": 6.8, "status": "Medical hold", "good_with_kids": True},
]
print(len(animals), "animals")
```

```python
print(animals[0]["name"])
print(animals[-1]["status"])
```

```python
print(f"{'ID':<8}{'Name':<9}{'Species':<12}{'Age':>5}  Status")
for animal in animals:
    print(f"{animal['id']:<8}{animal['name']:<9}"
          f"{animal['species']:<12}{animal['age']:>5}  {animal['status']}")
```

```python
available = []
for animal in animals:
    if animal["status"] == "Available":
        available.append(animal["name"])
print(available)
```

```python
family_friendly = [a["name"] for a in animals
                   if a["status"] == "Available" and a["good_with_kids"]]
print(family_friendly)
```

```python
oldest = animals[0]
for animal in animals:
    if animal["age"] > oldest["age"]:
        oldest = animal
print(f"Oldest: {oldest['name']}, {oldest['age']} years")
```

```python
total = 0
for animal in animals:
    if animal["status"] == "Medical hold":
        continue
    fee = fees[animal["species"]]
    if animal["species"] in ("Dog", "Cat") and animal["age"] >= 8:
        fee = fee / 2
    print(f"{animal['name']:<8} ${fee:>6.2f}")
    total += fee
print(f"{'Total':<8} ${total:>6.2f}")
```

```python
by_species = {}
for animal in animals:
    species = animal["species"]
    by_species[species] = by_species.get(species, 0) + 1
print(by_species)
```

```python
animals.append({"id": "SP-047", "name": "Nugget", "species": "Dog",
                "age": 2, "weight": 30.0, "status": "Available",
                "good_with_kids": True})
print(len(animals), animals[-1]["name"])
```

## 9.8 Nested data

```python
volunteers = {
    "Ken Adebayo": {"days": ["Tue", "Wed", "Sat"],
                    "skills": ["dog walking", "training"]},
    "Lucy Fairbanks": {"days": ["Sat", "Sun"],
                       "skills": ["cat care", "photography"]},
    "Tomas Novak": {"days": ["Tue", "Thu", "Sat"],
                    "skills": ["small animals", "cleaning"]},
    "Mei Chen": {"days": ["Tue", "Fri"],
                 "skills": ["front desk", "photography"]},
}
print(volunteers["Lucy Fairbanks"])
print(volunteers["Lucy Fairbanks"]["days"])
print(volunteers["Lucy Fairbanks"]["days"][0])
```

```python
for name, info in volunteers.items():
    if "Sat" in info["days"]:
        print(f"{name} works on Saturday")
```

```python
volunteers["Mei Chen"]["days"].append("Sun")
print(volunteers["Mei Chen"])
```

```python
shelter = {
    "name": "Sunny Paws Animal Rescue",
    "town": "Maple Falls",
    "opened": 2015,
    "hours": ("10:00", "17:00"),
    "spaces": {"dog kennels": 20, "cat condos": 16, "hutches": 8},
}
opens, closes = shelter["hours"]
print(f"{shelter['name']}, {shelter['town']} (since {shelter['opened']})")
print(f"Open {opens} to {closes}")
print(f"Total spaces: {sum(shelter['spaces'].values())}")
```

## 9.9 Sets: one of each

```python
species_seen = set(arrivals)
print(len(arrivals), "arrivals")
print(len(species_seen), "different species")
print(sorted(species_seen))
```

```python
kinds = {"Dog", "Cat"}
kinds.add("Rabbit")
kinds.add("Dog")
print(len(kinds))
print("Dog" in kinds, "Ferret" in kinds)
```

```python
for kind in sorted(kinds):
    print(kind)
```

```python
tuesday = {"Ken Adebayo", "Tomas Novak", "Mei Chen"}
saturday = {"Ken Adebayo", "Lucy Fairbanks", "Tomas Novak",
            "Aisha Bello"}
both = tuesday & saturday
either = tuesday | saturday
only_tuesday = tuesday - saturday
print("Both days:", ", ".join(sorted(both)))
print("Either day:", ", ".join(sorted(either)))
print("Tuesday only:", ", ".join(sorted(only_tuesday)))
```

```python
skills = set()
for info in volunteers.values():
    for skill in info["skills"]:
        skills.add(skill)
print(len(skills), "different skills:")
for skill in sorted(skills):
    print("-", skill)
```

```python
needed = {"dog walking", "cat care", "fundraising"}
missing = needed - skills
print("Nobody on the list can do:", missing)
```

```python
empty = {}
print(type(empty))
print(type(set()))
```
