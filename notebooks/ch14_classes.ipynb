## 14.3 Methods: giving objects behavior

```python
FEES = {"Dog": 150, "Cat": 90, "Rabbit": 45, "Guinea pig": 25}


class Animal:
    """One animal at Sunny Paws."""

    def __init__(self, animal_id, name, species, breed, age_years,
                 weight_kg, status="Available"):
        self.animal_id = animal_id
        self.name = name
        self.species = species
        self.breed = breed
        self.age_years = age_years
        self.weight_kg = weight_kg
        self.status = status

    def human_years(self):
        """Age in human years, using the shelter's poster rule."""
        if self.species == "Dog":
            per_year = 5
        elif self.species == "Cat":
            per_year = 4
        else:
            return None
        if self.age_years < 1:
            years = self.age_years * 15
        elif self.age_years < 2:
            years = 15 + (self.age_years - 1) * 9
        else:
            years = 15 + 9 + (self.age_years - 2) * per_year
        return round(years, 1)

    def is_senior(self):
        return self.species in ("Dog", "Cat") and self.age_years >= 8

    def adoption_fee(self):
        """The fee in dollars, or None if the animal can't be adopted."""
        if self.status == "Medical hold":
            return None
        fee = FEES[self.species]
        if self.is_senior():
            fee = fee // 2
        return fee

    def food_grams(self):
        """Grams of dry food (or pellets) per day."""
        if self.species == "Dog":
            return round(self.weight_kg * 15)
        if self.species == "Cat":
            if self.age_years < 1:
                return 80
            return 60
        if self.species == "Rabbit":
            return 80
        return 40
```

```python
peanut = Animal("SP-001", "Peanut", "Dog", "Labrador mix", 4.5, 28.6)
print(peanut.human_years())
print(peanut.adoption_fee())
print(peanut.food_grams())
```

## 14.4 Printing objects nicely, and changing them with methods

```python
class Animal:
    """One animal at Sunny Paws."""

    def __init__(self, animal_id, name, species, breed, age_years,
                 weight_kg, status="Available"):
        self.animal_id = animal_id
        self.name = name
        self.species = species
        self.breed = breed
        self.age_years = age_years
        self.weight_kg = weight_kg
        self.status = status
        self.adopted_on = ""

    def __str__(self):
        return f"{self.name} ({self.species}, {self.age_years} years)"

    def human_years(self):
        """Age in human years, using the shelter's poster rule."""
        if self.species == "Dog":
            per_year = 5
        elif self.species == "Cat":
            per_year = 4
        else:
            return None
        if self.age_years < 1:
            years = self.age_years * 15
        elif self.age_years < 2:
            years = 15 + (self.age_years - 1) * 9
        else:
            years = 15 + 9 + (self.age_years - 2) * per_year
        return round(years, 1)

    def is_senior(self):
        return self.species in ("Dog", "Cat") and self.age_years >= 8

    def is_adoptable(self):
        return self.status in ("Available", "Foster")

    def adoption_fee(self):
        """The fee in dollars, or None if the animal can't be adopted."""
        if not self.is_adoptable():
            return None
        fee = FEES[self.species]
        if self.is_senior():
            fee = fee // 2
        return fee

    def food_grams(self):
        """Grams of dry food (or pellets) per day."""
        if self.species == "Dog":
            return round(self.weight_kg * 15)
        if self.species == "Cat":
            if self.age_years < 1:
                return 80
            return 60
        if self.species == "Rabbit":
            return 80
        return 40

    def adopt(self, date):
        """Record an adoption on the given date, like "2026-07-04"."""
        if not self.is_adoptable():
            raise ValueError(f"{self.name} cannot be adopted "
                             f"({self.status})")
        self.status = "Adopted"
        self.adopted_on = date
```

## 14.5 Creating many objects from a CSV file

```python
import csv


def make_animal(row):
    """Turn one row from animals.csv (a dict of strings) into an Animal."""
    animal = Animal(row["id"], row["name"], row["species"], row["breed"],
                    float(row["age_years"]), float(row["weight_kg"]),
                    row["status"])
    animal.adopted_on = row["adopted_on"]
    return animal


animals = []
with open("../data/animals.csv", encoding="utf-8", newline="") as file:
    for row in csv.DictReader(file):
        animals.append(make_animal(row))

print(len(animals))
for animal in animals[:5]:
    print(animal)
```

```python
oldest = animals[0]
for animal in animals:
    if animal.age_years > oldest.age_years:
        oldest = animal
print(f"Oldest: {oldest}, about {oldest.human_years()} in human years")

seniors = [animal.name for animal in animals
           if animal.is_senior() and animal.is_adoptable()]
print("Seniors looking for a home:")
print(", ".join(seniors))
```

## 14.6 A Shelter class

```python
class Shelter:
    """A shelter holds a list of Animal objects."""

    def __init__(self, name):
        self.name = name
        self.animals = []

    def __str__(self):
        return f"{self.name} ({len(self.animals)} animals)"

    def add(self, animal):
        self.animals.append(animal)

    def load_csv(self, path):
        """Add every animal in a CSV file like data/animals.csv."""
        with open(path, encoding="utf-8", newline="") as file:
            for row in csv.DictReader(file):
                self.add(make_animal(row))

    def find(self, animal_id):
        """Return the animal with this id, or None."""
        for animal in self.animals:
            if animal.animal_id == animal_id:
                return animal
        return None

    def adoptable(self, species=None):
        """Animals that can be adopted, optionally of one species."""
        found = []
        for animal in self.animals:
            if animal.is_adoptable():
                if species is None or animal.species == species:
                    found.append(animal)
        return found

    def species_counts(self):
        counts = {}
        for animal in self.animals:
            counts[animal.species] = counts.get(animal.species, 0) + 1
        return counts
```

```python
sunny_paws = Shelter("Sunny Paws Animal Rescue")
sunny_paws.load_csv("../data/animals.csv")
print(sunny_paws)
print(sunny_paws.species_counts())
```

## 14.7 Inheritance: Dog and Cat

```python
class Animal:
    """One animal at Sunny Paws."""

    def __init__(self, animal_id, name, species, breed, age_years,
                 weight_kg, status="Available"):
        self.animal_id = animal_id
        self.name = name
        self.species = species
        self.breed = breed
        self.age_years = age_years
        self.weight_kg = weight_kg
        self.status = status
        self.adopted_on = ""

    def __str__(self):
        return f"{self.name} ({self.species}, {self.age_years} years)"

    def human_years(self):
        """The poster only has rules for dogs and cats."""
        return None

    def food_grams(self):
        """Grams of pellets per day for a rabbit or guinea pig."""
        if self.species == "Rabbit":
            return 80
        return 40

    def is_senior(self):
        return self.species in ("Dog", "Cat") and self.age_years >= 8

    def is_adoptable(self):
        return self.status in ("Available", "Foster")

    def adoption_fee(self):
        if not self.is_adoptable():
            return None
        fee = FEES[self.species]
        if self.is_senior():
            fee = fee // 2
        return fee

    def adopt(self, date):
        if not self.is_adoptable():
            raise ValueError(f"{self.name} cannot be adopted "
                             f"({self.status})")
        self.status = "Adopted"
        self.adopted_on = date
```

```python
class Dog(Animal):
    """A dog: uses the dog rules for human years and food."""

    def __init__(self, animal_id, name, breed, age_years, weight_kg,
                 status="Available"):
        super().__init__(animal_id, name, "Dog", breed, age_years,
                         weight_kg, status)

    def human_years(self):
        if self.age_years < 1:
            years = self.age_years * 15
        elif self.age_years < 2:
            years = 15 + (self.age_years - 1) * 9
        else:
            years = 15 + 9 + (self.age_years - 2) * 5
        return round(years, 1)

    def food_grams(self):
        return round(self.weight_kg * 15)


class Cat(Animal):
    """A cat: uses the cat rules for human years and food."""

    def __init__(self, animal_id, name, breed, age_years, weight_kg,
                 status="Available"):
        super().__init__(animal_id, name, "Cat", breed, age_years,
                         weight_kg, status)

    def human_years(self):
        if self.age_years < 1:
            years = self.age_years * 15
        elif self.age_years < 2:
            years = 15 + (self.age_years - 1) * 9
        else:
            years = 15 + 9 + (self.age_years - 2) * 4
        return round(years, 1)

    def food_grams(self):
        if self.age_years < 1:
            return 80
        return 60
```

```python
peanut = Dog("SP-001", "Peanut", "Labrador mix", 4.5, 28.6)
print(peanut)
print(peanut.species)
print(peanut.human_years())
print(peanut.adoption_fee())
print(isinstance(peanut, Dog), isinstance(peanut, Animal))
print(isinstance(peanut, Cat))
```

```python
def make_animal(row):
    """Turn one row from animals.csv into a Dog, Cat or Animal."""
    age = float(row["age_years"])
    weight = float(row["weight_kg"])
    if row["species"] == "Dog":
        animal = Dog(row["id"], row["name"], row["breed"], age, weight,
                     row["status"])
    elif row["species"] == "Cat":
        animal = Cat(row["id"], row["name"], row["breed"], age, weight,
                     row["status"])
    else:
        animal = Animal(row["id"], row["name"], row["species"],
                        row["breed"], age, weight, row["status"])
    animal.adopted_on = row["adopted_on"]
    return animal


sunny_paws = Shelter("Sunny Paws Animal Rescue")
sunny_paws.load_csv("../data/animals.csv")
for animal in sunny_paws.animals[:6]:
    print(f"{animal.name:<9}{animal.species:<12}"
          f"{str(animal.human_years()):<6}{animal.food_grams():>4} g")
```
