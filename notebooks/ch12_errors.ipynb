## 12.5 Catching the right errors

````python
def average_gift(total, count_text):
    """Return total / count, or None (with a message) if impossible."""
    try:
        count = int(count_text)
        return total / count
    except ValueError:
        print(f"'{count_text}' is not a whole number.")
    except ZeroDivisionError:
        print("No donors yet, so there is no average.")

print(average_gift(7805, "130"))
print(average_gift(7805, "lots"))
print(average_gift(0, "0"))
````

## 12.6 `else` and `finally`

````python
def read_age(text):
    """Try to read an age and report how it went."""
    try:
        age = float(text)
    except ValueError:
        print(f"Could not read '{text}'.")
    else:
        print(f"Got it: {age} years.")
    finally:
        print("Checked one answer.")

read_age("4.5")
print()
read_age("old")
````

## 12.7 A robust number-input function

````python
def ask_number(prompt, minimum=None, maximum=None):
    """Ask until the user types a number in range. Return it as a float.

    minimum and maximum are optional limits (both included).
    """
    while True:
        text = input(prompt).strip()
        try:
            number = float(text)
        except ValueError:
            print(f"Sorry, '{text}' is not a number. "
                  "Please type digits, like 4 or 4.5.")
            continue
        if minimum is not None and number < minimum:
            print(f"Please enter a number of at least {minimum}.")
        elif maximum is not None and number > maximum:
            print(f"Please enter a number no bigger than {maximum}.")
        else:
            return number
````

````python
age = ask_number("Age in years: ", minimum=0, maximum=30)
print("Saved age:", age)
````

````python
donation = ask_number("Donation in dollars: ", minimum=0)
print(f"Thank you for ${donation:.2f}!")
````

## 12.8 Raising your own errors

````python
def check_weight(weight_kg):
    """Return weight_kg if it is sensible; raise ValueError if not."""
    if weight_kg <= 0:
        raise ValueError(f"Weight must be over 0 kg, not {weight_kg}")
    if weight_kg > 100:
        raise ValueError(f"{weight_kg} kg is too heavy for a pet")
    return weight_kg

print(check_weight(28.6))
print(check_weight(-3))
````

````python
def strict_adoption_fee(species, age_years, status):
    """Return the adoption fee in dollars for one animal.

    Raises ValueError for an unknown species or an animal on
    medical hold, instead of returning None.
    """
    fees = {"Dog": 150, "Cat": 90, "Rabbit": 45, "Guinea pig": 25}
    if species not in fees:
        raise ValueError(f"Unknown species: {species}")
    if status == "Medical hold":
        raise ValueError("Animals on medical hold cannot be adopted")
    fee = fees[species]
    if species in ("Dog", "Cat") and age_years >= 8:
        fee = fee // 2
    return fee
````

````python
checks = [("Dog", 3, "Available"), ("Hamster", 1, "Available"),
          ("Dog", 0.9, "Medical hold")]
for species, age, status in checks:
    try:
        fee = strict_adoption_fee(species, age, status)
        print(f"{species}: ${fee}")
    except ValueError as error:
        print(f"{species}: {error}")
````

## 12.10 Debugging basics

### Print debugging

````python
def average_donation(amounts):
    """Return the average of a list of donation amounts."""
    total = 0
    for amount in amounts:
        total = amount
    return total / len(amounts)

print(average_donation([20, 50, 100, 30]))
````

````python
def average_donation(amounts):
    """Return the average of a list of donation amounts."""
    total = 0
    for amount in amounts:
        total = amount
        print(f"DEBUG: amount={amount}, total={total}")
    return total / len(amounts)

print(average_donation([20, 50, 100, 30]))
````

````python
def average_donation(amounts):
    """Return the average of a list of donation amounts."""
    total = 0
    for amount in amounts:
        total += amount
    return total / len(amounts)

print(average_donation([20, 50, 100, 30]))
````
