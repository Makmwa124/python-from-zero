## 17.3 Step 1: the rules

```python
from datetime import date

from rules import adoption_fee, human_years, is_adoption_day

print(human_years("Dog", 3), human_years("Cat", 3))
print(adoption_fee("Cat", 8, "Available"))
print(adoption_fee("Dog", 3, "Available", adoption_day=True))
print(is_adoption_day(date(2026, 10, 3)), is_adoption_day(date(2026, 10, 6)))
```

## 17.4 Step 2: the Animal and Shelter classes

```python
from animals import Animal, Shelter

kiwi = Animal("SP-014", "Kiwi", "Dog", "Beagle", "F", 3.0, 12.2,
              "2026-02-18", good_with_kids=True)
print(kiwi)
print(kiwi.human_years(), kiwi.fee(), kiwi.daily_food())
print(kiwi.matches("BEAGLE"), kiwi.matches("hamster"))
```

```python
shelter = Shelter()
shelter.add(kiwi)
print(shelter.find("sp-014").name)
print(shelter.next_id())
print(shelter.spaces_free("Dog"))
```

## 17.5 Step 3: loading and saving

```python
from pathlib import Path

from storage import DATA, load_shelter, save_shelter, start_shelter

print(DATA.name)
Path("output").mkdir(exist_ok=True)
save_file = Path("output/ch17_shelter_data.json")
shelter, message = start_shelter(save_file)
print(message)
```

```python
peanut = shelter.find("SP-001")
print(peanut)
print(type(peanut.age_years), type(peanut.good_with_kids))
```

```python
fee = shelter.adopt("SP-036", date(2026, 10, 6))
print(fee, shelter.find("SP-036").status)
save_shelter(shelter, save_file)
again = load_shelter(save_file)
print(again.find("SP-036").status, again.find("SP-036").adopted_on)
```
