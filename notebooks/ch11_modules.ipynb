## 11.4 Dates with `datetime`

```python
from datetime import date, timedelta

print(date.today())
```

```python
today = date(2026, 6, 30)
print(today)
print(today.year, today.month, today.day)
```

```python
arrived = date.fromisoformat("2026-01-07")
waiting = today - arrived
print(waiting)
print(waiting.days)
print(type(waiting))
```

```python
check_in = today + timedelta(days=14)
print("Foster check-in:", check_in)
print("Two weeks ago:", today - timedelta(weeks=2))
```

```python
print(today.weekday())
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday",
        "Saturday", "Sunday"]
print(days[today.weekday()])
```

```python
print(today.strftime("%A, %B %d, %Y"))
print(today.strftime("%d/%m/%Y"))
print(today.strftime("%a %d %b"))
```

```python
def is_open(day):
    """Return True if Sunny Paws is open on this date."""
    return day.weekday() != 0  # 0 means Monday

for offset in range(7):
    day = today + timedelta(days=offset)
    if is_open(day):
        status = "open"
    else:
        status = "closed"
    print(day.strftime("%a %b %d"), status)
```

```python
def adoption_day(year):
    """Return the date of Adoption Day: the first Saturday in October."""
    day = date(year, 10, 1)
    while day.weekday() != 5:  # 5 means Saturday
        day = day + timedelta(days=1)
    return day

for year in [2026, 2027, 2028]:
    print(year, adoption_day(year).strftime("%A, %B %d"))
```

## 11.5 Averages with `statistics`

```python
import statistics

ages = [4.5, 4.5, 3.5, 3, 14, 2, 11, 7.5, 1, 10, 2, 0.5]
print("Mean:  ", round(statistics.mean(ages), 1))
print("Median:", statistics.median(ages))
print("Mode:  ", statistics.mode(ages))
print("Spread:", round(statistics.stdev(ages), 1))
```

## 11.8 Sunny Paws tools: pet of the week and days waiting

```python
def pet_of_the_week(animals, day):
    """Pick one animal at random. The pick stays the same all week."""
    week = day.isocalendar().week
    random.seed(day.year * 100 + week)
    return random.choice(animals)

for offset in [0, 1, 3, 7]:
    day = today + timedelta(days=offset)
    pet = pet_of_the_week(ANIMALS, day)
    print(day, "week", day.isocalendar().week, pet["name"])
```

```python
def days_since(date_text, today):
    """Return the number of days from a "YYYY-MM-DD" date to today."""
    arrived = date.fromisoformat(date_text)
    return (today - arrived).days

print(days_since("2026-01-07", today))
print(days_since("2026-06-02", today))
```
