## 4.8 f-strings

```python
name = "Peanut"
species = "Dog"
age = 4.5
print(f"{name} is a {age}-year-old {species}.")
```

```python
print("Age: " + age)
```

```python
print(f"Age: {age}")
print(f"{name.upper()} has {len(name)} letters.")
print(f"In months, that is {age * 12} months.")
```

```python
fee = 150
print(f"Fee: ${fee}")
print(f"Fee: ${fee:.2f}")
weight = 28.6
per_day = weight * 15
print(f"Food per day: {per_day} g")
print(f"Food per day: {per_day:.0f} g")
```

```python
total_donations = 7805
print(f"Donations so far: ${total_donations:,}")
print(f"Donations so far: ${total_donations:,.2f}")
print(f"Average gift: ${total_donations / 130:.2f}")
```

```python
name = "Kiwi"
print(f"[{name:<10}]")
print(f"[{name:>10}]")
print(f"[{name:^10}]")
```

```python
dog_fee = 150
cat_fee = 90
rabbit_fee = 45
guinea_pig_fee = 25
print(f"{'Species':<12}{'Fee ($)':>10}")
print("-" * 22)
print(f"{'Dog':<12}{dog_fee:>10.2f}")
print(f"{'Cat':<12}{cat_fee:>10.2f}")
print(f"{'Rabbit':<12}{rabbit_fee:>10.2f}")
print(f"{'Guinea pig':<12}{guinea_pig_fee:>10.2f}")
```

```python
print(f"{'Sunny Paws':*^30}")
print(f"{' Adoption Day ':=^30}")
```

## 4.9 Project: a name tag and a kennel card

```python
name = "Toast"
species = "Cat"
breed = "Maine Coon mix"
age = 3.5

print("+" + "-" * 28 + "+")
print(f"|{name.upper():^28}|")
print(f"|{species + ', ' + breed:^28}|")
print(f"|{'Age: ' + str(age) + ' years':^28}|")
print("+" + "-" * 28 + "+")
```

```python
record = "SP-014,Kiwi,Dog,Beagle,F,3,12.2,2026-02-18,Available"
kennel = 7

animal_id = record[:6]
name = "Kiwi"
breed = "Beagle"
weight = 12.2
arrived = record[-20:-10]

print("=" * 32)
print(f" KENNEL {kennel:<10}{animal_id:>13}")
print("=" * 32)
print(f" {'Name:':<10}{name}")
print(f" {'Breed:':<10}{breed}")
print(f" {'Weight:':<10}{weight:.1f} kg")
print(f" {'Arrived:':<10}{arrived}")
print(f" {'Food:':<10}{weight * 15:.0f} g dry food a day")
print("=" * 32)
```

```python
line = "2026-06-03 13:29 |  PRIYA SHAH |  asked about Rosie"
first_bar = line.find("|")
second_bar = line.find("|", first_bar + 1)
print(first_bar)
print(second_bar)
```

```python
date = line[:10]
visitor = line[first_bar + 1:second_bar].strip().title()
reason = line[second_bar + 1:].strip()
print(f"{date}  {visitor:<14}{reason}")
```
