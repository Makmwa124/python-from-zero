## 10.5 The shelter's three functions

### `human_years()`

```python
def human_years(species, age_years):
    """Return an animal's age in human years, using the shelter's poster.

    Works for dogs and cats. Returns None for any other species,
    because the poster has no rule for them.
    """
    if species == "Dog":
        per_later_year = 5
    elif species == "Cat":
        per_later_year = 4
    else:
        return None

    if age_years < 1:
        years = age_years * 15
    elif age_years < 2:
        years = 15 + (age_years - 1) * 9
    else:
        years = 15 + 9 + (age_years - 2) * per_later_year
    return round(years, 1)
```

```python
print(human_years("Dog", 3))
print(human_years("Cat", 3))
print(human_years("Dog", 0.4))
print(human_years("Cat", 1.5))
print(human_years("Dog", 4.5))
print(human_years("Rabbit", 3))
```

### `adoption_fee()`

```python
def adoption_fee(species, age_years, status):
    """Return the adoption fee in dollars for one animal.

    Dogs and cats aged 8 or older (seniors) pay half price.
    Returns None if the animal is on medical hold (it cannot be
    adopted yet) or if the species is not one the shelter takes.
    """
    if status == "Medical hold":
        return None
    fees = {"Dog": 150, "Cat": 90, "Rabbit": 45, "Guinea pig": 25}
    if species not in fees:
        return None
    fee = fees[species]
    if species in ("Dog", "Cat") and age_years >= 8:
        fee = fee // 2
    return fee
```

```python
print(adoption_fee("Dog", 4.5, "Available"))
print(adoption_fee("Dog", 10, "Foster"))
print(adoption_fee("Cat", 14, "Available"))
print(adoption_fee("Rabbit", 3, "Available"))
print(adoption_fee("Dog", 0.9, "Medical hold"))
print(adoption_fee("Hamster", 1, "Available"))
```

```python
fee = adoption_fee("Dog", 0.9, "Medical hold")
if fee is None:
    print("Olive cannot be adopted yet.")
else:
    print(f"Olive's fee is ${fee}.")
```

### `food_grams()`

```python
def food_grams(species, age_years, weight_kg):
    """Return grams of dry food (or pellets) per day for one animal.

    Dogs: 15 g per kg of body weight, rounded. Cats: 60 g, or 80 g
    for kittens under 1 year. Rabbits: 80 g of pellets. Guinea pigs:
    40 g of pellets. Returns None for any other species.
    """
    if species == "Dog":
        return round(weight_kg * 15)
    elif species == "Cat":
        if age_years < 1:
            return 80
        return 60
    elif species == "Rabbit":
        return 80
    elif species == "Guinea pig":
        return 40
    return None
```

```python
print(food_grams("Dog", 4.5, 28.6))
print(food_grams("Cat", 0.6, 1.7))
print(food_grams("Cat", 14, 3.9))
print(food_grams("Rabbit", 3, 2.5))
print(food_grams("Guinea pig", 2, 1.1))
```

### Using all three on a list of animals

```python
animals = [
    {"name": "Peanut", "species": "Dog", "age_years": 4.5,
     "weight_kg": 28.6, "status": "Available"},
    {"name": "Waffles", "species": "Cat", "age_years": 0.6,
     "weight_kg": 1.7, "status": "Foster"},
    {"name": "Poppy", "species": "Dog", "age_years": 10,
     "weight_kg": 30.7, "status": "Foster"},
    {"name": "Hazel", "species": "Rabbit", "age_years": 3,
     "weight_kg": 2.5, "status": "Foster"},
    {"name": "Ollie", "species": "Cat", "age_years": 14,
     "weight_kg": 3.9, "status": "Available"},
    {"name": "Olive", "species": "Dog", "age_years": 0.9,
     "weight_kg": 6.8, "status": "Medical hold"},
    {"name": "Pip", "species": "Guinea pig", "age_years": 2,
     "weight_kg": 1.1, "status": "Foster"},
]
```

```python
for animal in animals:
    species = animal["species"]
    age = animal["age_years"]
    fee = adoption_fee(species, age, animal["status"])
    grams = food_grams(species, age, animal["weight_kg"])
    if fee is None:
        fee_text = "not available"
    else:
        fee_text = f"${fee}"
    print(f"{animal['name']:<8} {species:<11} {grams:>4} g/day  "
          f"fee: {fee_text}")
```

## 10.6 Default values and keyword arguments

```python
def feeding_note(name, grams, meals=2):
    per_meal = round(grams / meals)
    return f"{name}: {grams} g a day, {meals} meals of {per_meal} g"

print(feeding_note("Peanut", 429))
print(feeding_note("Olive", 102, 3))
```

```python
print(feeding_note(name="Waffles", grams=80, meals=4))
print(feeding_note("Waffles", meals=4, grams=80))
print(feeding_note("Waffles", 80, meals=4))
```

## 10.7 Returning more than one value

```python
def feeding_plan(species, age_years, weight_kg):
    """Return (grams per day, how to feed) for one animal."""
    grams = food_grams(species, age_years, weight_kg)
    if species == "Dog":
        if age_years < 1:
            how = "three times a day"
        else:
            how = "morning and evening"
    elif species == "Cat":
        if age_years < 1:
            how = "four small meals"
        else:
            how = "morning and evening"
    elif species == "Rabbit":
        how = "pellets, plus unlimited hay"
    else:
        how = "pellets, plus hay and fresh veg"
    return grams, how
```

```python
grams, how = feeding_plan("Dog", 0.9, 6.8)
print(f"Olive: {grams} g, {how}")

plan = feeding_plan("Rabbit", 3, 2.5)
print(plan)
print(type(plan))
```

```python
def weight_range(weights):
    """Return the lightest and heaviest weight in a list."""
    return min(weights), max(weights)

lightest, heaviest = weight_range([28.6, 1.7, 30.7, 2.5, 3.9, 6.8])
print(f"From {lightest} kg to {heaviest} kg")
```
